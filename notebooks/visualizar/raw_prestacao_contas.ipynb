```python
from pathlib import Path
import duckdb

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pyproject.toml").exists())
CSV = ROOT / "dados/raw/prestacao_contas/2026/prestacao_contas_candidatos_2026/receitas_candidatos_2026_BRASIL.csv"
con = duckdb.connect()
con.execute("LOAD encodings")
con.sql(f"CREATE OR REPLACE VIEW dados_raw AS SELECT * FROM read_csv('{CSV.as_posix()}', delim=';', header=true, encoding='cp1252', all_varchar=true, ignore_errors=true)")
con.sql(f"COPY dados_raw TO '{ROOT / "dados" / "processed" / "prestacao_contas.parquet"}' (FORMAT PARQUET, COMPRESSION ZSTD)")
con.sql(f"CREATE VIEW dados AS SELECT * FROM '{ROOT / "dados" / "processed" / "prestacao_contas.parquet"}'")
```

```python
con.sql("DESCRIBE dados").df()
```

```python
con.sql("SUMMARIZE dados").df()
```

```python
con.sql("SELECT COUNT(*) FROM dados").df() #NUM DE COLUNAS
```
